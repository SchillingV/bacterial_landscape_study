```python
for string_feature in dict_dia_charts.keys():

    chart_dia_three_config_comparison = go.Figure()
    chart_dia_four_config_comparison = go.Figure()

    for string_configuration in dict_dia_configs.keys():

        df_dia = pd.read_csv(
            dict_dia_configs[string_configuration]['path'], delimiter='\t')
        chart_dia = go.Figure()

        if string_feature in ["MS1_Accuracy_Comparison", "MS2_Accuracy_Comparison"]:
            chart_dia.add_trace(
                go.Violin(
                    x=["value"] * len(df_dia),
                    y=df_dia[dict_dia_charts[string_feature]["Column_Name"][0]],
                    name=dict_dia_charts[string_feature]["Column_Name"][0],
                    box_visible=True,
                    points="all",
                    pointpos=-1.2,      # points to the left
                    jitter=0.05,
                    side="negative",
                    line_color=dict_dia_configs[string_configuration]['color'],
                    marker_color=dict_dia_configs[string_configuration]['color']
                )
            )

            chart_dia.add_trace(
                go.Violin(
                    x=["value"] * len(df_dia),
                    y=df_dia[dict_dia_charts[string_feature]["Column_Name"][1]],
                    name=dict_dia_charts[string_feature]["Column_Name"][1],
                    box_visible=True,
                    points="all",
                    pointpos=1.2,      # points to the right
                    jitter=0.05,
                    side="positive",
                    line_color=dict_dia_configs[string_configuration]['color_strong'],
                    marker_color=dict_dia_configs[string_configuration]['color_strong']
                )
            )

            chart_dia.update_layout(
                violinmode="overlay",
                violingap=0,
                height=500,
                width=700,
                margin=dict(l=80, r=20, t=20, b=80),
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="right",
                    x=1)
            )
            chart_dia.update_yaxes(title=dict(text=string_feature, font=dict(size=30)),
            tickfont=dict(size=25), showgrid=True)
            chart_dia.update_xaxes(
                showticklabels=False, ticks="", showline=True,
                tickfont=dict(size=25))

        elif string_feature == 'Peak_Analysis':
            chart_dia.add_trace(
                go.Scatter(
                    x=df_dia["FWHM.RT"], y=df_dia["Total.Quantity"],
                    mode='markers'
                )
            )
            chart_dia.update_traces(
                line_color=dict_dia_configs[string_configuration]['color'],
                marker_color=dict_dia_configs[string_configuration]['color'])

            chart_dia.update_xaxes(showgrid=True,
                title=dict(font=dict(size=30)), tickfont=dict(size=25))
            chart_dia.update_yaxes(showgrid=True,
                title=dict(font=dict(size=30)), tickfont=dict(size=25))
            chart_dia.update_layout(
                margin=dict(l=80, r=20, t=20, b=80))

        else:
            chart_dia.add_trace(
                go.Violin(
                    x=["value"] * len(df_dia),
                    y=df_dia[dict_dia_charts[string_feature]["Column_Name"]],
                    name=dict_dia_charts[string_feature]["Column_Name"],
                    box_visible=True,
                    points="all",
                    line_color=dict_dia_configs[string_configuration]['color'],
                    marker_color=dict_dia_configs[string_configuration]['color']
                )
            )

            chart_dia.update_xaxes(
                showticklabels=False, ticks="", showline=True,
                tickfont=dict(size=25))
            chart_dia.update_yaxes(
                showgrid=True,
                title=dict(text=dict_dia_charts[string_feature]["Column_Name"],
                    font=dict(size=30)),
                tickfont=dict(size=25))
            chart_dia.update_layout(
                margin=dict(l=80, r=20, t=20, b=80))

            # We also build a chart that compares a single feature across
            # either three or four configurations (100SPD, 200SPD, ...).
            # We do this only for singular features, and not feature
            # combinations that create more advanced plots
            if string_configuration != 'all':
                chart_dia_three_config_comparison.add_trace(
                    go.Violin(
                        x=[string_configuration]*len(
                            df_dia[dict_dia_charts[string_feature]["Column_Name"]]),
                        y=df_dia[dict_dia_charts[string_feature]["Column_Name"]],
                        name=dict_dia_charts[string_feature]["Column_Name"],
                        box_visible=True,
                        points="all",
                        line_color=dict_dia_configs[string_configuration]['color'],
                        marker_color=dict_dia_configs[string_configuration]['color']
                    )
                )

            chart_dia_four_config_comparison.add_trace(
                go.Violin(
                    x=[string_configuration]*len(
                        df_dia[dict_dia_charts[string_feature]["Column_Name"]]),
                    y=df_dia[dict_dia_charts[string_feature]["Column_Name"]],
                    name=dict_dia_charts[string_feature]["Column_Name"],
                    box_visible=True,
                    points="all",
                    line_color=dict_dia_configs[string_configuration]['color'],
                    marker_color=dict_dia_configs[string_configuration]['color']
                )
            )

        # We check if the four comparison figure is empty or not
        if chart_dia_three_config_comparison.data != tuple():
            chart_dia_three_config_comparison.update_yaxes(
                showgrid=True,
                title=dict(text=dict_dia_charts[string_feature]["Column_Name"],
                    font=dict(size=30)),
                tickfont=dict(size=25))
            chart_dia_three_config_comparison.update_xaxes(
                tickfont=dict(size=25))
            chart_dia_three_config_comparison.update_layout(
                margin=dict(l=80, r=20, t=20, b=80),
                violingap=0.5,
                height=500,
                width=800,
                showlegend=False)

            chart_dia_three_config_comparison.write_image(
                os.path.join(
                    _PATH_RESULTS_DIA_PDF,
                    "DIA_Three_Config_Comparison_" + string_feature + ".pdf"))
            chart_dia_three_config_comparison.write_html(
                os.path.join(
                    _PATH_RESULTS_DIA_HTML,
                    "DIA_Three_Config_Comparison_" + string_feature + ".html"))
            dict_dia_charts[string_feature]["Charts"][
                'three_comparison'] = chart_dia_three_config_comparison

        # We check if the four comparison figure is empty or not
        if chart_dia_four_config_comparison.data != tuple():
            chart_dia_four_config_comparison.update_yaxes(
                showgrid=True,
                title=dict(text=dict_dia_charts[string_feature]["Column_Name"],
                    font=dict(size=30)),
                tickfont=dict(size=25))
            chart_dia_four_config_comparison.update_xaxes(
                tickfont=dict(size=25))
            chart_dia_four_config_comparison.update_layout(
                margin=dict(l=80, r=20, t=20, b=80),
                violingap=0.5,
                height=500,
                width=1000,
                showlegend=False)

            chart_dia_four_config_comparison.write_image(
                os.path.join(
                    _PATH_RESULTS_DIA_PDF,
                    "DIA_Four_Config_Comparison_" + string_feature + ".pdf"))
            chart_dia_four_config_comparison.write_html(
                os.path.join(
                    _PATH_RESULTS_DIA_HTML,
                    "DIA_Four_Config_Comparison_" + string_feature + ".html"))
            dict_dia_charts[string_feature]["Charts"][
                'four_comparison'] = chart_dia_four_config_comparison

        chart_dia.write_image(
            os.path.join(
                _PATH_RESULTS_DIA_PDF,
                "DIA_" + string_configuration + "_" + string_feature + ".pdf"))
        chart_dia.write_html(
            os.path.join(
                _PATH_RESULTS_DIA_HTML,
                "DIA_" + string_configuration + "_" + string_feature + ".html"))

        dict_dia_charts[string_feature]["Charts"][string_configuration] = chart_dia
```

```python
# =====================================================================
# Comparison: 100SPD-DIA vs 100SPD-DDA identification metrics
# One paired-violin figure per metric, in the same style as the
# three-config DIA comparison plots.
# =====================================================================

# Load the two 100SPD tables. df_dda is already in memory from the DDA
# block; we (re)load the 100SPD DIA table explicitly to be self-contained.
df_dia_100spd = pd.read_csv(dict_dia_configs['100SPD']['path'], delimiter='\t')

# Metric -> (DIA column, DDA column) for the genuinely comparable counts.
# All four are reported per run; DIA uses DIA-NN column names, DDA uses
# the FragPipe/QC column names.
dict_dia_dda_comparison = {
    'Precursors_Identified': {
        'dia_col': 'Precursors.Identified',
        'dda_col': 'Identified_precursors',
    },
    'Peptides_Identified': {
        'dia_col': 'Peptides.Identified',
        'dda_col': 'Identified_peptides',
    },
    'Proteins_Identified': {
        'dia_col': 'Proteins.Identified',
        'dda_col': 'Identified_proteins',
    },
    'Protein_Identification_Rate': {
        'dia_col': 'Protein.Identification.Rate',
        'dda_col': 'Protein.Identification.Rate',
    },
}

# x-axis group label -> (dataframe, column-key, color) is resolved per metric.
dict_comparison_modes = {
    '100SPD-DIA': {
        'df': df_dia_100spd,
        'color': _STRING_COLOR_DIA100SPD,
    },
    '100SPD-DDA': {
        'df': df_dda,
        'color': _STRING_COLOR_DARK_GREY,
    },
}

for string_feature in dict_dia_dda_comparison.keys():
    print(string_feature)

    chart_dia_dda_comparison = go.Figure()

    for string_mode in dict_comparison_modes.keys():

        df_mode = dict_comparison_modes[string_mode]['df']

        # pick the right column name for this mode
        if string_mode == '100SPD-DIA':
            column_name = dict_dia_dda_comparison[string_feature]['dia_col']
        else:
            column_name = dict_dia_dda_comparison[string_feature]['dda_col']

        chart_dia_dda_comparison.add_trace(
            go.Violin(
                x=[string_mode] * len(df_mode[column_name]),
                y=df_mode[column_name],
                name=string_mode,
                box_visible=True,
                points="all",
                line_color=dict_comparison_modes[string_mode]['color'],
                marker_color=dict_comparison_modes[string_mode]['color']
            )
        )

    chart_dia_dda_comparison.update_yaxes(
        showgrid=True,
        title=dict(text=string_feature, font=dict(size=30)),
        tickfont=dict(size=25))
    chart_dia_dda_comparison.update_xaxes(
        tickfont=dict(size=25))
    chart_dia_dda_comparison.update_layout(
        margin=dict(l=80, r=20, t=20, b=80),
        violingap=0.5,
        height=500,
        width=800,
        showlegend=False)

    chart_dia_dda_comparison.write_image(
        os.path.join(
            _PATH_RESULTS_DIA_PDF,
            "DIA_DDA_100SPD_Comparison_" + string_feature + ".pdf"))
    chart_dia_dda_comparison.write_html(
        os.path.join(
            _PATH_RESULTS_DIA_HTML,
            "DIA_DDA_100SPD_Comparison_" + string_feature + ".html"))
```
